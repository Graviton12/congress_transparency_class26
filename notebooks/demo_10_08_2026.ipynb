# Pandas Skill: many-to-many joins (usually we don't want this! but here we do)

```python
vote_compare = pd.merge(votes, votes, 
                         on = ['chamber', 'rollnumber']) # merge votes with itself on chamber and rollnumber
vote_compare # automatically specifies which icpsr and cast_code coems from the left side and the right side
```

# Pandas Skill: more dropping

```python
vote_compare = vote_compare.drop(['chamber', 'rollnumber'], axis=1) # remove the chmaber and rollnumber columns since we dont need them anymore
vote_compare
```

# Pandas Skill: create new column

```python
vote_compare['agree'] = vote_compare['cast_code_x'] == vote_compare['cast_code_y'] # create a new column 'agree' to indicate whether the cast codes match
vote_compare
```

```python
vote_compare = vote_compare.drop(['cast_code_x', 'cast_code_y'], axis=1) # remove the cast_code columns since we dont need them anymore
vote_compare
```

# Pandas Skill: filter rows

```python
vote_compare = vote_compare.query('icpsr_x != icpsr_y') # filter out rows where the icpsr values are the same
vote_compare
```

# Pandas Skill: rename columns

```python
vote_compare = vote_compare.rename({'icpsr_x': 'icpsr', 
                                    'icpsr_y': 'comparison member'},
                                    axis=1)
vote_compare
```

# Pandas Skill: sorting

```python
vote_compare = vote_compare.sort_values(['icpsr', 'comparison member'], ascending=True) # sort the dataframe by icpsr and comparison member columns in ascending order
vote_compare
```

# Pandas Skill: data aggregation

```python
vote_compare = vote_compare.groupby(['icpsr', 'comparison member']).agg({'agree': 'mean'})
vote_compare
```

```python
vote_compare = vote_compare.groupby(['icpsr', 'comparison member']).agg({'agree': 'mean'}).reset_index()
vote_compare
```
