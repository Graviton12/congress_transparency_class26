There are five pieces of information you need to naviagte any API and its poor documentation:
1. How you get an API key (or another credentiial) to get access to the data (check)

2. Root: the beginning of the URl for any endpooint contained within the API

A URL: `root / endpoint ? paramters`

3. Endpoint: the middle of the URL, specifies which data store on the API to access

4. Parameters: ways to control what exactly is returned when you access an endpoint. There are two kinds of paramters:
    * query paramters: appear after the ?, but requests formats this for using the `params` arguement
    * path parameters: supposed to be entered as part of the endpoint itself
    
5. How to supply your API key back tot he API to prove that you are you

```python
# style: I dont like putting a slash at the end of the root because the slash is usually part of the endpoint
root =  'https://api.congress.gov/v3' 

endpoint = '/member'

params = {'format': 'json', 'currentMember': 'true', 'api_key': congresskey}
```

```python
r = requests.get(root + endpoint, params=params)
r
```
