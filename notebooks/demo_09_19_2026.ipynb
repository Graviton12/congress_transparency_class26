There are five pieces of information you need to naviagte any API and its poor documentation:
1. How you get an API key (or another credentiial) to get access to the data (check)

2. Root: the beginning of the URl for any endpooint contained within the API

A URL: `root / endpoint ? parameters`

3. Endpoint: the middle of the URL, specifies which data store on the API to access

4. Parameters (End point parameters): ways to control what exactly is returned when you access an endpoint. There are two kinds of paramters:
    * query paramters: appear after the ?, but requests formats this for using the `params` arguement
    * path parameters: supposed to be entered as part of the endpoint itself
    
5. How to supply your API key back tot he API to prove that you are you

```python
# we see that teh default limit is 20 (20 congress people) but we want the max 250 congress people
params['limit'] = '250'
```

```python
r = requests.get(root + endpoint, params=params, headers=headers)
myjson = r.json()
congress_df = pd.json_normalize(myjson, record_path=['members'])

pd.set_option('display.max_rows', 10)
congress_df
```

```python
params['offset'] = '500'
r = requests.get(root + endpoint, params=params, headers=headers)
myjson = r.json()
congress_df = pd.json_normalize(myjson, record_path=['members'])
congress_df
```

```python
i = 0 
still_members = True
congress_json = []
while still_members:
    print(i)
    params['offset'] = i
    r = requests.get(root + endpoint, params=params, headers=headers)
    myjson = r.json()
    still_members = len(myjson['members']) > 0
    congress_json.extend(myjson['members'])
    i = i + 250
```
